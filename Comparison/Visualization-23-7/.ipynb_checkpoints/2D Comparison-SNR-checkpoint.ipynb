```python
import os
import tifffile
import stackview
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from skimage import exposure
```

```python
time = 50
attn = "0.7"
```

```python
root_dir = "/nfs/scratch2/shared_image_recog_ml/spatial_dino_exp/ap2_unnorm"
folders = [
    f for f in os.listdir(root_dir)
    if os.path.isdir(os.path.join(root_dir, f))
    and not f.startswith(".")
]
folders.sort()

selected_folder = folders[time-1]

volume_path = os.path.join(root_dir, selected_folder, "volume.tif")
volume = tifffile.imread(volume_path)
```

```python
CME = pd.read_csv("Input/detections_CME.csv", header = None)
Dino = pd.read_csv(f"Input/ap2_july22_attn_{attn}.csv", names = ["t","ID", "z", "y", "x"],skiprows=1)
Dino = Dino[["t","ID","x","y","z"]]
Comp = pd.read_csv(f"Input/DC_attn{attn}.csv")

Dino["t"] = Dino["t"]+1

Dino_time = Dino[Dino["t"]==time]
CME_time = CME[CME.iloc[:, 0]==time]

Comp_col = Comp[Comp["Distance"]<3.5]
Comp_col_time = Comp_col[Comp_col.iloc[:, 0]==time]

Comp_col_time["Multi ID (CME)"] = Comp_col_time["Multi ID (CME)"].str.split(',')
Comp_col_time["Multi Distance (CME)"] = Comp_col_time["Multi Distance (CME)"].str.split(',')
Comp_col_time = Comp_col_time.explode(["Multi ID (CME)", "Multi Distance (CME)"])

Dino_ID_col = Comp_col_time["ID (Dino)"].unique()
CME_ID_col = Comp_col_time["Multi ID (CME)"].unique().astype(int)

Dino_clean = Dino_time[~Dino_time["ID"].isin(Dino_ID_col)].reset_index(drop=True)
CME_clean = CME_time[~CME_time[1].isin(CME_ID_col)].reset_index(drop=True)
```

```python
max_projection = np.max(volume, axis=0)

plt.figure(figsize=(35,23))

plt.subplot(2,3,1)
plt.imshow(max_projection, cmap="grey",vmin=0, vmax=0.5)
plt.scatter(CME_time.iloc[:,:][2],CME_time.iloc[:,:][3], s = 8,color = "blue", label ="CME")  # All CME detections for a time point
plt.scatter(Dino_time.iloc[:,:]["x"],Dino_time.iloc[:,:]["y"], s = 8, color = "red", label = "Dino")  # All Dino detections for a time point
plt.axis("off")
plt.legend(markerscale=3,fontsize = 15,loc=2)
plt.title("All Dino Detections & All CME Detections", fontsize=20)

plt.subplot(2,3,2)
plt.imshow(max_projection, cmap="grey",vmin=0, vmax=0.5)
plt.scatter(CME_time.iloc[:,:][2],CME_time.iloc[:,:][3], s = 8,color = "blue", label ="CME")  # All CME detections for a time point
plt.scatter(Dino_time.iloc[:,:]["x"],Dino_time.iloc[:,:]["y"], s = 8, color = "red", label = "Dino")  # All Dino detections for a time point
plt.scatter((Comp_col_time.iloc[:,:]["x (CME)"]+Comp_col_time.iloc[:,:]["x (Dino)"])/2,(Comp_col_time.iloc[:,:]["y (CME)"]+Comp_col_time.iloc[:,:]["y (Dino)"])/2, s = 8, color = "green", label = "Colocalization") # matched detections for a time point
plt.axis("off")
plt.legend(markerscale=3,fontsize =15,loc=2)
plt.title("All Detections & Colocalization", fontsize=20)

plt.subplot(2,3,3)
plt.imshow(max_projection, cmap="grey",vmin=0, vmax=0.5)
plt.scatter((Comp_col_time.iloc[:,:]["x (CME)"]+Comp_col_time.iloc[:,:]["x (Dino)"])/2,(Comp_col_time.iloc[:,:]["y (CME)"]+Comp_col_time.iloc[:,:]["y (Dino)"])/2, s = 8, color = "green", label = "Colocalization") # matched detections for a time point
plt.axis("off")
plt.legend(markerscale=3,fontsize =15,loc=2)
plt.title("Colocalization", fontsize=20)


plt.subplot(2,3,4)
plt.imshow(max_projection, cmap="grey",vmin=0, vmax=0.5)
plt.scatter(Dino_clean.iloc[:,:]["x"],Dino_clean.iloc[:,:]["y"], s = 8, color = "red", label = "Dino")  # Dino detections with matches removed for at time point
plt.axis("off")
plt.legend(markerscale=3,fontsize =15,loc=2)
plt.title("Dino Non-Colocalization", fontsize=20)

plt.subplot(2,3,5)
plt.imshow(max_projection, cmap="grey",vmin=0, vmax=0.5)
plt.scatter(CME_clean.iloc[:,:][2],CME_clean.iloc[:,:][3], s = 8,color = "blue", label = "CME")  # CME detections with matches removed for at time point
plt.scatter(Dino_clean.iloc[:,:]["x"],Dino_clean.iloc[:,:]["y"], s = 8, color = "red", label = "Dino")  # Dino detections with matches removed for at time point
plt.axis("off")
plt.legend(markerscale=3,fontsize =15,loc=2)
plt.title("Non-Colocalization", fontsize=20)

plt.subplot(2,3,6)
plt.imshow(max_projection, cmap="grey",vmin=0, vmax=0.5)
plt.scatter(CME_clean.iloc[:,:][2],CME_clean.iloc[:,:][3], s = 8,color = "blue", label = "CME")  # CME detections with matches removed for at time point
plt.axis("off")
plt.legend(markerscale=3,fontsize =15,loc=2)
plt.title("CME Non-Colocalization", fontsize=20)

plt.suptitle(f"attn {attn} t = {time}", fontsize = 25)
plt.tight_layout()
plt.savefig(f"attn{attn}t{time}.png")
plt.show()
```
